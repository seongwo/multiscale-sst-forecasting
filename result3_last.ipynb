```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class FactorizedSTBlock(nn.Module):
    def __init__(self, d_model, nhead, mlp_ratio=4.0, dropout=0.0, attn_dropout=0.0):
        super().__init__()
        self.ln_s = nn.LayerNorm(d_model)
        self.attn_s = nn.MultiheadAttention(
            d_model, nhead, dropout=attn_dropout, batch_first=True
        )

        self.ln_t = nn.LayerNorm(d_model)
        self.attn_t = nn.MultiheadAttention(
            d_model, nhead, dropout=attn_dropout, batch_first=True
        )

        self.ln_m = nn.LayerNorm(d_model)
        hidden = int(d_model * mlp_ratio)
        self.mlp = nn.Sequential(
            nn.Linear(d_model, hidden),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden, d_model),
            nn.Dropout(dropout),
        )

    def forward(
        self,
        x,
        return_attn: bool = False,
        average_attn_weights: bool = False,
    ):
        """
        x: (B, T, N, D)

        return_attn=False:
            기존과 동일하게 x만 반환

        return_attn=True:
            x, aux 반환
            aux = {
                "spatial_attn": (B, T, H, N, N) or (B, T, N, N),
                "temporal_attn": (B, N, H, T, T) or (B, N, T, T),
            }
        """
        B, T, N, D = x.shape
        aux = {}

        # Spatial attention: (B*T, N, D)
        xs = self.ln_s(x).reshape(B * T, N, D)
        if return_attn:
            ys, attn_s = self.attn_s(
                xs, xs, xs,
                need_weights=True,
                average_attn_weights=average_attn_weights,
            )
        else:
            ys, _ = self.attn_s(xs, xs, xs, need_weights=False)

        x = x + ys.reshape(B, T, N, D)

        # Temporal attention: (B*N, T, D)
        xt = self.ln_t(x).permute(0, 2, 1, 3).reshape(B * N, T, D)
        if return_attn:
            yt, attn_t = self.attn_t(
                xt, xt, xt,
                need_weights=True,
                average_attn_weights=average_attn_weights,
            )
        else:
            yt, _ = self.attn_t(xt, xt, xt, need_weights=False)

        x = x + yt.reshape(B, N, T, D).permute(0, 2, 1, 3)

        x = x + self.mlp(self.ln_m(x))

        if return_attn:
            # attn_s:
            # average_attn_weights=False -> (B*T, num_heads, N, N)
            # average_attn_weights=True  -> (B*T, N, N)
            if average_attn_weights:
                attn_s = attn_s.reshape(B, T, N, N)
            else:
                Hh = attn_s.shape[1]
                attn_s = attn_s.reshape(B, T, Hh, N, N)

            # attn_t:
            # average_attn_weights=False -> (B*N, num_heads, T, T)
            # average_attn_weights=True  -> (B*N, T, T)
            if average_attn_weights:
                attn_t = attn_t.reshape(B, N, T, T)
            else:
                Hh = attn_t.shape[1]
                attn_t = attn_t.reshape(B, N, Hh, T, T)

            aux["spatial_attn"] = attn_s.detach()
            aux["temporal_attn"] = attn_t.detach()
            return x, aux

        return x


class MSFormerV3(nn.Module):
    def __init__(
        self,
        image_size=32,
        in_chans=1,
        tin=14,
        tout=7,
        d_model=128,
        depth=4,
        num_heads=4,
        mlp_ratio=4.0,
        dropout=0.0,
        attn_dropout=0.0,
        spatial_scales=(2, 4, 8, 16, 32),
        fusion="sum",          # "sum", "gated", "gated2"
        use_pad=True,
        use_pos_emb=False,
        time_pool="last",      # "mean" or "last"
    ):
        super().__init__()
        self.image_size = int(image_size)
        self.in_chans = int(in_chans)
        self.tin = int(tin)
        self.tout = int(tout)
        self.d_model = int(d_model)
        self.depth = int(depth)

        self.scales = tuple(int(s) for s in spatial_scales)
        assert len(self.scales) >= 1
        self.fusion = fusion
        assert self.fusion in ["sum", "gated", "gated2"]
        self.use_pad = bool(use_pad)
        self.use_pos_emb = bool(use_pos_emb)
        self.time_pool = str(time_pool)
        assert self.time_pool in ["mean", "last"]

        self.smoother = nn.Sequential(
            nn.Conv2d(self.in_chans, self.in_chans, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv2d(self.in_chans, self.in_chans, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv2d(self.in_chans, self.in_chans, kernel_size=3, padding=1),
            nn.GELU(),
            nn.Conv2d(self.in_chans, self.in_chans, kernel_size=3, padding=1),
        )

        self.tokenizers = nn.ModuleDict()
        for s in self.scales:
            self.tokenizers[str(s)] = nn.Conv2d(
                self.in_chans, self.d_model, kernel_size=s, stride=s, bias=True
            )

        self.time_emb = nn.Embedding(self.tin, self.d_model)
        self.scale_emb = nn.Embedding(len(self.scales), self.d_model)

        self.pos_emb = nn.ModuleDict()
        if self.use_pos_emb:
            for s in self.scales:
                gs = self.image_size // s
                Ns = gs * gs
                self.pos_emb[str(s)] = nn.Embedding(Ns, self.d_model)

        self.blocks = nn.ModuleList([
            FactorizedSTBlock(self.d_model, num_heads, mlp_ratio, dropout, attn_dropout)
            for _ in range(depth)
        ])
        self.ln_f = nn.LayerNorm(self.d_model)

        self.future_heads = nn.ModuleDict()
        self.token_to_patch = nn.ModuleDict()
        for s in self.scales:
            self.future_heads[str(s)] = nn.Linear(self.d_model, self.tout * self.d_model)
            self.token_to_patch[str(s)] = nn.Linear(self.d_model, self.in_chans * s * s)

        if self.fusion in ["gated", "gated2"]:
            self.gate = nn.Sequential(
                nn.Linear(self.d_model * len(self.scales), self.d_model),
                nn.GELU(),
                nn.Linear(self.d_model, len(self.scales)),
            )
        else:
            self.gate = None

    def _pad_to_multiple(self, x, multiple: int):
        h, w = x.shape[-2], x.shape[-1]
        pad_h = (multiple - (h % multiple)) % multiple
        pad_w = (multiple - (w % multiple)) % multiple
        if pad_h == 0 and pad_w == 0:
            return x, (0, 0)
        x = F.pad(x, (0, pad_w, 0, pad_h), mode="replicate")
        return x, (pad_h, pad_w)

    def _tokenize_one_scale(self, x_btchw, s: int):
        tok = self.tokenizers[str(s)](x_btchw)
        gh, gw = tok.shape[-2], tok.shape[-1]
        tok = tok.flatten(2).transpose(1, 2)   # (B*T, Ns, D)
        return tok, gh, gw

    def _decode_scale(self, fut, s: int, gh: int, gw: int):
        """
        fut: (B, Tout, Ns, D)
        return: (B, Tout, C, gh*s, gw*s)
        """
        B = fut.shape[0]
        patch_vec = self.token_to_patch[str(s)](fut)  # (B, Tout, Ns, C*s*s)
        patch_vec = patch_vec.reshape(B, self.tout, gh, gw, self.in_chans, s, s)
        patch_vec = patch_vec.permute(0, 1, 4, 2, 5, 3, 6).contiguous()
        y = patch_vec.reshape(B, self.tout, self.in_chans, gh * s, gw * s)
        return y

    def _pool_scale_rep(self, tok):
        # tok: (B, Tin, Ns, D)
        if self.time_pool == "last":
            rep = tok[:, -1].mean(dim=1)   # (B, D)
        else:
            rep = tok.mean(dim=(1, 2))     # (B, D)
        return rep

    def forward(
        self,
        x,
        return_interpret: bool = False,
        return_attn: bool = False,
        attn_layers=None,
        average_attn_weights: bool = True,
    ):
        """
        기본:
            y = model(x)

        해석:
            y, interpret = model(
                x,
                return_interpret=True,
                return_attn=True,
                attn_layers=[-1],   # 예: 마지막 블록만
                average_attn_weights=True,
            )
        """
        B, Tin, C, H, W = x.shape
        assert C == self.in_chans
        assert Tin == self.tin, f"Tin mismatch: got {Tin}, expected {self.tin}"

        if attn_layers is None:
            attn_layers = []
        attn_layers = set([(self.depth + i) if i < 0 else i for i in attn_layers])

        interpret = {} if return_interpret else None

        xt = x.reshape(B * Tin, C, H, W)
        if self.use_pad:
            xt, (pad_h, pad_w) = self._pad_to_multiple(xt, max(self.scales))
        else:
            pad_h, pad_w = 0, 0

        # -------------------------------------------------
        # tokenize + emb
        # -------------------------------------------------
        scale_tokens = []
        scale_meta = []
        t_idx = torch.arange(Tin, device=x.device)

        for sid, s in enumerate(self.scales):
            tok_s, gh, gw = self._tokenize_one_scale(xt, s)
            Ns = gh * gw
            tok_s = tok_s.reshape(B, Tin, Ns, self.d_model)

            tok_s = tok_s + self.time_emb(t_idx).view(1, Tin, 1, self.d_model)
            sid_tensor = torch.tensor(sid, device=x.device)
            tok_s = tok_s + self.scale_emb(sid_tensor).view(1, 1, 1, self.d_model)

            if self.use_pos_emb:
                p_idx = torch.arange(Ns, device=x.device)
                tok_s = tok_s + self.pos_emb[str(s)](p_idx).view(1, 1, Ns, self.d_model)

            scale_tokens.append(tok_s)
            scale_meta.append((s, Ns, gh, gw))

        if return_interpret:
            interpret["scale_token_shapes"] = {
                int(s): (int(B), int(Tin), int(Ns), int(self.d_model))
                for (s, Ns, _, _) in scale_meta
            }

        # -------------------------------------------------
        # pre-transformer gate
        # -------------------------------------------------
        if self.fusion == "gated":
            reps = [self._pool_scale_rep(tok_s) for tok_s in scale_tokens]
            cat = torch.cat(reps, dim=-1)
            w = torch.softmax(self.gate(cat), dim=-1)
        else:
            w = None

        if return_interpret:
            interpret["fusion_mode"] = self.fusion
            if w is not None:
                interpret["scale_weights"] = {
                    int(s): w[:, i].detach().cpu()
                    for i, (s, _, _, _) in enumerate(scale_meta)
                }

        # -------------------------------------------------
        # transformer
        # -------------------------------------------------
        z = torch.cat(scale_tokens, dim=2)  # (B, Tin, N_total, D)
        h = z

        if return_interpret and return_attn:
            interpret["spatial_attn"] = {}
            interpret["temporal_attn"] = {}

        for bi, blk in enumerate(self.blocks):
            if return_interpret and return_attn and (bi in attn_layers):
                h, attn_aux = blk(
                    h,
                    return_attn=True,
                    average_attn_weights=average_attn_weights,
                )
                interpret["spatial_attn"][bi] = attn_aux["spatial_attn"].cpu()
                interpret["temporal_attn"][bi] = attn_aux["temporal_attn"].cpu()
            else:
                h = blk(h)

        h = self.ln_f(h)

        # -------------------------------------------------
        # post-transformer gate
        # -------------------------------------------------
        if self.fusion == "gated2":
            reps_h = []
            cursor = 0
            for _, Ns, _, _ in scale_meta:
                tok_h = h[:, :, cursor:cursor + Ns, :]
                cursor += Ns
                reps_h.append(self._pool_scale_rep(tok_h))

            cat_h = torch.cat(reps_h, dim=-1)
            w = torch.softmax(self.gate(cat_h), dim=-1)

            if return_interpret:
                interpret["scale_weights"] = {
                    int(s): w[:, i].detach().cpu()
                    for i, (s, _, _, _) in enumerate(scale_meta)
                }

        # -------------------------------------------------
        # scale-wise decode
        # -------------------------------------------------
        ys = []
        cursor = 0

        if return_interpret:
            interpret["scale_outputs"] = {}
            interpret["scale_reps"] = {}
            interpret["scale_future_tokens"] = {}

        for i, (s, Ns, gh, gw) in enumerate(scale_meta):
            tok = h[:, :, cursor:cursor + Ns, :]
            cursor += Ns

            rep = tok[:, -1, :, :] if self.time_pool == "last" else tok.mean(dim=1)

            fut = self.future_heads[str(s)](rep).reshape(B, Ns, self.tout, self.d_model)
            fut = fut.permute(0, 2, 1, 3).contiguous()   # (B, Tout, Ns, D)

            y_s = self._decode_scale(fut, s, gh, gw)

            if w is not None:
                y_s = y_s * w[:, i].view(B, 1, 1, 1, 1)

            ys.append(y_s)

            if return_interpret:
                interpret["scale_reps"][int(s)] = rep.detach().cpu()
                interpret["scale_future_tokens"][int(s)] = fut.detach().cpu()
                interpret["scale_outputs"][int(s)] = y_s.detach().cpu()

        y = torch.stack(ys, dim=0).sum(dim=0)

        if return_interpret:
            interpret["pre_smoother"] = y.detach().cpu()

        if self.use_pad and (pad_h > 0 or pad_w > 0):
            y = y[..., :H, :W]

            if return_interpret:
                interpret["pre_smoother"] = interpret["pre_smoother"][..., :H, :W]
                for s in self.scales:
                    interpret["scale_outputs"][int(s)] = interpret["scale_outputs"][int(s)][..., :H, :W]

        # -------------------------------------------------
        # smoother
        # -------------------------------------------------
        B2, T2, C2, H2, W2 = y.shape
        y2 = y.reshape(B2 * T2, C2, H2, W2)
        smooth_delta = self.smoother(y2)
        y2 = y2 + smooth_delta
        y = y2.reshape(B2, T2, C2, H2, W2)

        if return_interpret:
            interpret["smoother_delta"] = smooth_delta.reshape(B2, T2, C2, H2, W2).detach().cpu()
            interpret["post_smoother"] = y.detach().cpu()

        if return_interpret:
            return y, interpret
        return y


class MSFv3(nn.Module):
    def __init__(self, **kwargs):
        super().__init__()
        kwargs.setdefault("fusion", "sum")
        self.net = MSFormerV3(**kwargs)

    def forward(self, x, **kwargs):
        return self.net(x, **kwargs)


class MSFGv3(nn.Module):
    def __init__(self, **kwargs):
        super().__init__()
        kwargs["fusion"] = "gated"
        self.net = MSFormerV3(**kwargs)

    def forward(self, x, **kwargs):
        return self.net(x, **kwargs)


class MSFG2v3(nn.Module):
    def __init__(self, **kwargs):
        super().__init__()
        kwargs["fusion"] = "gated2"
        self.net = MSFormerV3(**kwargs)

    def forward(self, x, **kwargs):
        return self.net(x, **kwargs)
```

```python
def scale_boundaries_from_shapes(scale_token_shapes):
    """
    scale_token_shapes: interpret["scale_token_shapes"]
    예:
      {
        2: (B, Tin, Ns, D),
        4: (B, Tin, Ns, D),
        ...
      }

    return:
      {scale: (start, end)}
    """
    bounds = {}
    cursor = 0
    for s in sorted(scale_token_shapes.keys()):
        _, _, Ns, _ = scale_token_shapes[s]
        bounds[s] = (cursor, cursor + Ns)
        cursor += Ns
    return bounds
```

```python

MISSING_VALUE = -9.96921e36

class OISSTDatasetWithMeta(Dataset):
    def __init__(
        self,
        ds,
        mask: np.ndarray,
        mean: float,
        std: float,
        tin: int = 14,
        tout: int = 7,
        var_name: str = "sst",
    ):
        self.tin = int(tin)
        self.tout = int(tout)
        self.mask = mask.astype(bool)
        self.mean = float(mean)
        self.std = float(std) if float(std) != 0.0 else 1e-12

        arr = ds[var_name].values.astype(np.float32)
        self.arr = arr
        self.time = ds.time.values

        self.T = arr.shape[0]
        self.H = arr.shape[1]
        self.W = arr.shape[2]

        if self.mask.shape != (self.H, self.W):
            raise ValueError(f"mask shape {self.mask.shape} != data spatial shape {(self.H, self.W)}")

        self.n = self.T - self.tin - self.tout + 1
        if self.n <= 0:
            raise ValueError("Not enough timesteps for the requested (tin, tout).")

    def __len__(self):
        return self.n

    def __getitem__(self, idx: int):
        seq = self.arr[idx : idx + self.tin + self.tout]
        x = seq[: self.tin]
        y = seq[self.tin :]

        x = (x - self.mean) / self.std
        y = (y - self.mean) / self.std

        ocean = self.mask[None, :, :]
        x = np.where(ocean, x, 0.0)
        y = np.where(ocean, y, 0.0)

        x = torch.from_numpy(x).unsqueeze(1)
        y = torch.from_numpy(y).unsqueeze(1)

        t_ref = self.time[idx + self.tin]
        t_py = np.datetime_as_string(t_ref, unit="D")
        year, month, day = map(int, t_py.split("-"))

        jan1 = np.datetime64(f"{year}-01-01")
        doy = int((np.datetime64(t_py) - jan1).astype("timedelta64[D]").astype(int)) + 1

        meta = {
            "idx": int(idx),
            "t_ref": t_py,
            "year": int(year),
            "month": int(month),
            "doy": int(doy),
        }
        return x, y, meta


def open_oisst(path: str | Path):
    path = str(path)
    if path.endswith(".zarr"):
        return xr.open_zarr(path)
    return xr.open_dataset(path)


def compute_mean_std_train(ds_train: xr.Dataset, mask: np.ndarray, var_name: str = "sst"):
    arr = ds_train[var_name].values.astype(np.float32)
    valid = np.isfinite(arr) & (arr != MISSING_VALUE)
    valid &= mask[None, :, :]
    x = arr[valid].astype(np.float64)
    return float(x.mean()), float(x.std() + 1e-12)


def build_test_dataset(
    data_path: str | Path,
    mask: np.ndarray,
    tin: int = 14,
    tout: int = 7,
    var_name: str = "sst",
    train_start: str = "1983-01-01",
    train_end: str = "2015-12-31",
    test_start: str = "2021-01-01",
    test_end: str = "2024-12-31",
    stats_cache_json: str | Path | None = None,
):
    mask = mask.astype(bool)
    ds = open_oisst(data_path)
    ds_train = ds.sel(time=slice(train_start, train_end))
    ds_test = ds.sel(time=slice(test_start, test_end))

    mean, std = compute_mean_std_train(ds_train, mask, var_name=var_name)

    if stats_cache_json is not None:
        p = Path(stats_cache_json)
        p.parent.mkdir(parents=True, exist_ok=True)
        with p.open("w", encoding="utf-8") as f:
            json.dump({"mean": mean, "std": std}, f, indent=2)

    test_set = OISSTDatasetWithMeta(
        ds_test, mask, mean, std, tin=tin, tout=tout, var_name=var_name
    )
    torch_mask = torch.from_numpy(mask).bool()
    return test_set, torch_mask, mean, std
```

```python

def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def ensure_5d_batched(x: torch.Tensor) -> torch.Tensor:
    if x.dim() == 5:
        return x
    if x.dim() == 4:
        return x.unsqueeze(2)
    raise ValueError(f"Expected 4D/5D batched tensor, got {tuple(x.shape)}")


def ensure_pred_5d(pred: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
    if pred.dim() == 5:
        out = pred
    elif pred.dim() == 4:
        out = pred.unsqueeze(2)
    else:
        raise ValueError(f"Expected pred 4D/5D, got {tuple(pred.shape)}")

    if out.shape[:2] != y.shape[:2] or out.shape[-2:] != y.shape[-2:]:
        raise ValueError(f"pred shape mismatch: pred={tuple(out.shape)} vs y={tuple(y.shape)}")
    if out.shape[2] != y.shape[2]:
        raise ValueError(f"channel mismatch: pred C={out.shape[2]} vs y C={y.shape[2]}")
    return out


def denorm(x, mean, std):
    return x * std + mean


def show_sst_grid(
    images,
    titles=None,
    ncols=4,
    figsize=(16, 4),
    vmin=None,
    vmax=None,
    mask=None,
    flipud=True,
    cmap_name="coolwarm",
):
    n = len(images)
    nrows = math.ceil(n / ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=figsize)
    axes = np.array(axes).reshape(-1)

    cmap = plt.cm.get_cmap(cmap_name).copy()
    cmap.set_bad(color="black")

    for ax in axes[n:]:
        ax.axis("off")

    for i, img in enumerate(images):
        ax = axes[i]
        arr = np.asarray(img)

        if flipud:
            arr = np.flipud(arr)

        if mask is not None:
            m = np.asarray(mask).astype(bool)
            if flipud:
                m = np.flipud(m)
            arr = np.ma.masked_where(~m, arr)

        im = ax.imshow(arr, vmin=vmin, vmax=vmax, cmap=cmap, origin="upper")

        if titles is not None:
            ax.set_title(titles[i], fontsize=10)
        ax.axis("off")

        plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

    plt.tight_layout()
    plt.show()

    
def scale_boundaries(scale_meta):
    bounds = {}
    cursor = 0
    for s, info in scale_meta.items():
        Ns = info["Ns"]
        bounds[int(s)] = (cursor, cursor + Ns)
        cursor += Ns
    return bounds


def query_index_from_hw(h, w, gh, gw, H=32, W=32):
    ph = min(int(h * gh / H), gh - 1)
    pw = min(int(w * gw / W), gw - 1)
    return ph * gw + pw
```

```python

@torch.no_grad()
def eval_test(model, loader, mask_hw, device, eps_mape=1e-3):
    model.eval()
    sums = {"mse": 0.0, "rmse": 0.0, "mae": 0.0, "mape": 0.0, "r2": 0.0}
    rmse_t_sum = None
    mae_t_sum = None
    n = 0

    for x, y, meta in loader:
        x = ensure_5d_batched(x.to(device, non_blocking=True))
        y = ensure_5d_batched(y.to(device, non_blocking=True))

        pred = model(x)
        pred = ensure_pred_5d(pred, y)

        sums["mse"] += masked_mse(pred, y, mask_hw).item()
        sums["rmse"] += masked_rmse(pred, y, mask_hw).item()
        sums["mae"] += masked_mae(pred, y, mask_hw).item()
        sums["mape"] += masked_mape(pred, y, mask_hw, eps=eps_mape, as_percent=True).item()
        sums["r2"] += masked_r2(pred, y, mask_hw).item()

        rmse_t = masked_rmse_per_t(pred, y, mask_hw).detach().cpu()
        mae_t = masked_mae_per_t(pred, y, mask_hw).detach().cpu()

        if rmse_t_sum is None:
            rmse_t_sum = rmse_t.clone()
            mae_t_sum = mae_t.clone()
        else:
            rmse_t_sum += rmse_t
            mae_t_sum += mae_t
        n += 1

    for k in sums:
        sums[k] /= max(n, 1)

    sums["rmse_t"] = (rmse_t_sum / max(n, 1)).tolist() if rmse_t_sum is not None else None
    sums["mae_t"] = (mae_t_sum / max(n, 1)).tolist() if mae_t_sum is not None else None
    return sums


test_metrics = eval_test(model, test_loader, mask_hw, device, eps_mape=cfg["eps_mape"])
test_metrics
```

```python

@torch.no_grad()
def get_interpret_batch(
    model,
    loader,
    device,
    batch_index=0,
    attn_layers=[-1],
    average_attn_weights=True,
):
    model.eval()
    for bi, (x, y, meta) in enumerate(loader):
        if bi != batch_index:
            continue
        x = ensure_5d_batched(x.to(device, non_blocking=True))
        y = ensure_5d_batched(y.to(device, non_blocking=True))

        pred, interpret = model(
            x,
            return_interpret=True,
            return_attn=True,
            attn_layers=attn_layers,
            average_attn_weights=average_attn_weights,
        )
        pred = ensure_pred_5d(pred, y)
        return {
            "x": x.detach().cpu(),
            "y": y.detach().cpu(),
            "pred": pred.detach().cpu(),
            "meta": meta,
            "interpret": interpret,
        }
    raise IndexError("batch_index out of range")


bundle = get_interpret_batch(
    model=model,
    loader=test_loader,
    device=device,
    batch_index= 4,
    attn_layers=[-1],
    average_attn_weights=True,
)
bundle["pred"].shape, bundle["y"].shape
```

```python
import numpy as np
import pandas as pd
import torch

# =========================================================
# 전체 test set을 돌면서
# 1) sample별 예측 오차
# 2) sample별 GT 구조성(front/gradient/variation)
# 3) scale output 크기
# 를 모아서 "논문용 대표 sample" 후보를 고르는 코드
# =========================================================

def _broadcast_mask_5d(mask_hw: torch.Tensor, x: torch.Tensor) -> torch.Tensor:
    """
    mask_hw: (H, W) bool or float
    x: (B, T, C, H, W)
    return: (B, T, C, H, W) float
    """
    if mask_hw.dtype == torch.bool:
        m = mask_hw.float()
    else:
        m = mask_hw
    while m.dim() < x.dim():
        m = m.unsqueeze(0)
    return m.expand_as(x).to(device=x.device, dtype=x.dtype)


def _per_sample_rmse(pred: torch.Tensor, target: torch.Tensor, mask_hw: torch.Tensor, eps: float = 1e-8):
    """
    pred, target: (B, T, C, H, W)
    return:
      rmse_sample: (B,)
      rmse_lead:   (B, T)
    """
    m = _broadcast_mask_5d(mask_hw, pred)
    diff2 = (pred - target) ** 2

    num_bt = (diff2 * m).sum(dim=(2, 3, 4))                 # (B, T)
    den_bt = m.sum(dim=(2, 3, 4)).clamp_min(eps)            # (B, T)
    mse_bt = num_bt / den_bt
    rmse_bt = torch.sqrt(mse_bt.clamp_min(0.0))             # (B, T)

    num_b = (diff2 * m).sum(dim=(1, 2, 3, 4))               # (B,)
    den_b = m.sum(dim=(1, 2, 3, 4)).clamp_min(eps)          # (B,)
    mse_b = num_b / den_b
    rmse_b = torch.sqrt(mse_b.clamp_min(0.0))               # (B,)

    return rmse_b.detach().cpu().numpy(), rmse_bt.detach().cpu().numpy()


def _per_sample_mae(pred: torch.Tensor, target: torch.Tensor, mask_hw: torch.Tensor, eps: float = 1e-8):
    """
    pred, target: (B, T, C, H, W)
    return:
      mae_sample: (B,)
    """
    m = _broadcast_mask_5d(mask_hw, pred)
    diff = (pred - target).abs()
    num_b = (diff * m).sum(dim=(1, 2, 3, 4))
    den_b = m.sum(dim=(1, 2, 3, 4)).clamp_min(eps)
    mae_b = num_b / den_b
    return mae_b.detach().cpu().numpy()


def _masked_sample_spatial_std(y: torch.Tensor, mask_hw: torch.Tensor, eps: float = 1e-8):
    """
    y: (B, T, C, H, W)
    return:
      std_sample: (B,)  # lead별 std를 구해 평균
    """
    m = _broadcast_mask_5d(mask_hw, y)
    wsum = m.sum(dim=(2, 3, 4)).clamp_min(eps)                       # (B, T)
    mean_bt = (y * m).sum(dim=(2, 3, 4)) / wsum                     # (B, T)
    var_bt = (((y - mean_bt[:, :, None, None, None]) ** 2) * m).sum(dim=(2, 3, 4)) / wsum
    std_bt = torch.sqrt(var_bt.clamp_min(0.0))                      # (B, T)
    std_b = std_bt.mean(dim=1)                                      # (B,)
    return std_b.detach().cpu().numpy()


def _masked_sample_gradient_strength(y: torch.Tensor, mask_hw: torch.Tensor, eps: float = 1e-8):
    """
    y: (B, T, C, H, W)
    return:
      grad_sample: (B,)  # lead별 gradient magnitude 평균
    """
    m = _broadcast_mask_5d(mask_hw, y)

    dx = y[..., :, 1:] - y[..., :, :-1]                             # (B,T,C,H,W-1)
    mx = m[..., :, 1:] * m[..., :, :-1]
    gx_bt = (dx.abs() * mx).sum(dim=(2, 3, 4)) / mx.sum(dim=(2, 3, 4)).clamp_min(eps)

    dy = y[..., 1:, :] - y[..., :-1, :]                             # (B,T,C,H-1,W)
    my = m[..., 1:, :] * m[..., :-1, :]
    gy_bt = (dy.abs() * my).sum(dim=(2, 3, 4)) / my.sum(dim=(2, 3, 4)).clamp_min(eps)

    g_bt = 0.5 * (gx_bt + gy_bt)                                    # (B,T)
    g_b = g_bt.mean(dim=1)                                          # (B,)
    return g_b.detach().cpu().numpy()


def _per_sample_scale_mean_abs(scale_outputs_dict: dict):
    """
    scale_outputs_dict[s]: (B, T, C, H, W)
    return:
      out[s] = (B,)   # sample별 mean(abs(y_s))
    """
    out = {}
    for s, y_s in scale_outputs_dict.items():
        if torch.is_tensor(y_s):
            z = y_s.detach().cpu().numpy()
        else:
            z = np.asarray(y_s)
        out[int(s)] = np.mean(np.abs(z), axis=(1, 2, 3, 4))
    return out


@torch.no_grad()
def collect_sample_selection_table(
    model,
    loader,
    mask_hw,
    device,
    return_interpret=True,
):
    """
    전체 test set을 한 번 돌며 sample별 통계를 수집.
    attention은 안 뽑고 scale_outputs만 수집해서 속도/메모리 절약.

    return:
      df: sample별 요약 표
    """
    model.eval()

    rows = []
    global_row = 0

    for batch_idx, (x, y, meta) in enumerate(loader):
        x = ensure_5d_batched(x.to(device, non_blocking=True))
        y = ensure_5d_batched(y.to(device, non_blocking=True))

        if return_interpret:
            pred, interpret = model(
                x,
                return_interpret=True,
                return_attn=False,
            )
        else:
            pred = model(x)
            interpret = None

        pred = ensure_pred_5d(pred, y)

        rmse_b, rmse_bt = _per_sample_rmse(pred, y, mask_hw)
        mae_b = _per_sample_mae(pred, y, mask_hw)
        gt_std_b = _masked_sample_spatial_std(y, mask_hw)
        gt_grad_b = _masked_sample_gradient_strength(y, mask_hw)

        if interpret is not None and "scale_outputs" in interpret:
            scale_abs = _per_sample_scale_mean_abs(interpret["scale_outputs"])
        else:
            scale_abs = {}

        batch_size = x.shape[0]

        for i in range(batch_size):
            row = {
                "global_order": int(global_row),
                "batch_index": int(batch_idx),
                "sample_in_batch": int(i),
                "dataset_idx": int(meta["idx"][i]),
                "t_ref": str(meta["t_ref"][i]),
                "year": int(meta["year"][i]),
                "month": int(meta["month"][i]),
                "doy": int(meta["doy"][i]),

                "rmse_mean": float(rmse_b[i]),
                "mae_mean": float(mae_b[i]),
                "gt_spatial_std": float(gt_std_b[i]),
                "gt_gradient_strength": float(gt_grad_b[i]),
            }

            for t in range(rmse_bt.shape[1]):
                row[f"rmse_d{t+1}"] = float(rmse_bt[i, t])

            for s, vals in scale_abs.items():
                row[f"scale{s}_mean_abs"] = float(vals[i])

            rows.append(row)
            global_row += 1

    df = pd.DataFrame(rows)

    # season label 추가
    def month_to_season(m):
        if m in [12, 1, 2]:
            return "DJF"
        elif m in [3, 4, 5]:
            return "MAM"
        elif m in [6, 7, 8]:
            return "JJA"
        else:
            return "SON"

    df["season"] = df["month"].map(month_to_season)

    # 대표성 점수용 정규화
    rmse_med = df["rmse_mean"].median()
    df["rmse_distance_from_median"] = (df["rmse_mean"] - rmse_med).abs()

    # rank 기반 점수: 작을수록 좋음
    df["rank_rmse_repr"] = df["rmse_distance_from_median"].rank(method="average", ascending=True)
    df["rank_grad"] = df["gt_gradient_strength"].rank(method="average", ascending=False)
    df["rank_std"] = df["gt_spatial_std"].rank(method="average", ascending=False)

    # 대표성 + 구조성 종합 점수
    # - rmse는 중앙값 근처일수록 유리
    # - gradient/std가 클수록 front/structure가 더 잘 보임
    df["sample_score"] = (
        0.50 * df["rank_rmse_repr"] +
        0.30 * df["rank_grad"] +
        0.20 * df["rank_std"]
    )

    df = df.sort_values("sample_score", ascending=True).reset_index(drop=True)
    return df


def pick_representative_samples(
    df: pd.DataFrame,
    n=10,
    season=None,
    month=None,
    min_grad_quantile=0.50,
    min_std_quantile=0.50,
):
    """
    대표 sample 후보 추리기
    - rmse는 median 근처
    - gt gradient / gt std는 일정 이상
    """
    work = df.copy()

    if season is not None:
        work = work[work["season"] == season].copy()
    if month is not None:
        work = work[work["month"] == month].copy()

    if len(work) == 0:
        raise ValueError("조건에 맞는 sample이 없습니다.")

    g_thr = work["gt_gradient_strength"].quantile(min_grad_quantile)
    s_thr = work["gt_spatial_std"].quantile(min_std_quantile)

    work = work[
        (work["gt_gradient_strength"] >= g_thr) &
        (work["gt_spatial_std"] >= s_thr)
    ].copy()

    if len(work) == 0:
        raise ValueError("구조성 조건을 만족하는 sample이 없습니다. quantile을 낮춰보세요.")

    work = work.sort_values("sample_score", ascending=True).reset_index(drop=True)
    return work.head(n)


@torch.no_grad()
def get_interpret_for_global_sample(
    model,
    loader,
    device,
    global_order: int,
    return_attn=False,
    attn_layers=[-1],
    average_attn_weights=True,
):
    """
    전체 test set 기준 global_order번째 sample 하나를 interpret까지 포함해 가져오기.
    collect_sample_selection_table()의 global_order 컬럼과 매칭됨.
    """
    model.eval()
    cur = 0

    for batch_idx, (x, y, meta) in enumerate(loader):
        bsz = x.shape[0]
        if not (cur <= global_order < cur + bsz):
            cur += bsz
            continue

        local_idx = global_order - cur

        x = ensure_5d_batched(x.to(device, non_blocking=True))
        y = ensure_5d_batched(y.to(device, non_blocking=True))

        pred, interpret = model(
            x,
            return_interpret=True,
            return_attn=return_attn,
            attn_layers=attn_layers,
            average_attn_weights=average_attn_weights,
        )
        pred = ensure_pred_5d(pred, y)

        out = {
            "x": x[local_idx:local_idx+1].detach().cpu(),
            "y": y[local_idx:local_idx+1].detach().cpu(),
            "pred": pred[local_idx:local_idx+1].detach().cpu(),
            "meta": {
                "idx": [int(meta["idx"][local_idx])],
                "t_ref": [str(meta["t_ref"][local_idx])],
                "year": [int(meta["year"][local_idx])],
                "month": [int(meta["month"][local_idx])],
                "doy": [int(meta["doy"][local_idx])],
            },
            "interpret": {},
        }

        for k, v in interpret.items():
            if isinstance(v, dict):
                out["interpret"][k] = {}
                for kk, vv in v.items():
                    if torch.is_tensor(vv):
                        out["interpret"][k][kk] = vv[local_idx:local_idx+1].detach().cpu()
                    else:
                        out["interpret"][k][kk] = vv
            elif torch.is_tensor(v):
                out["interpret"][k] = v[local_idx:local_idx+1].detach().cpu()
            else:
                out["interpret"][k] = v

        return out

    raise IndexError("global_order가 test set 범위를 벗어났습니다.")


# =========================================================
# 사용 예시
# =========================================================

# 1) 전체 test set을 한 번 돌면서 sample별 표 생성
sample_df = collect_sample_selection_table(
    model=model,
    loader=test_loader,
    mask_hw=mask_hw,
    device=device,
    return_interpret=True,
)

print("전체 sample 수:", len(sample_df))
display_cols = [
    "global_order", "dataset_idx", "t_ref", "season", "month",
    "rmse_mean", "gt_spatial_std", "gt_gradient_strength", "sample_score"
]
print(sample_df[display_cols].head(15))

# 2) 전체 test set 기준 대표 sample 후보 10개
cands_all = pick_representative_samples(
    sample_df,
    n=10,
    min_grad_quantile=0.50,
    min_std_quantile=0.50,
)
print("\n[전체 test set 기준 대표 후보]")
print(cands_all[display_cols].to_string(index=False))

# 3) 계절별 대표 sample 후보도 확인 가능
for seas in ["DJF", "MAM", "JJA", "SON"]:
    try:
        cands_season = pick_representative_samples(
            sample_df,
            n=5,
            season=seas,
            min_grad_quantile=0.50,
            min_std_quantile=0.50,
        )
        print(f"\n[{seas} 대표 후보]")
        print(cands_season[display_cols].to_string(index=False))
    except Exception as e:
        print(f"\n[{seas}] 후보 선택 실패:", e)

# 4) 최종적으로 1개 선택해서 interpret bundle 가져오기
#    예: 전체 대표 후보 중 1번째
chosen = cands_all.iloc[0]
chosen_global_order = int(chosen["global_order"])

print("\n선택된 sample")
print(chosen[display_cols])

bundle_selected = get_interpret_for_global_sample(
    model=model,
    loader=test_loader,
    device=device,
    global_order=chosen_global_order,
    return_attn=False,   # scale output figure용이면 False로 충분
)

print("\n선택 bundle shape")
print("x:", bundle_selected["x"].shape)
print("y:", bundle_selected["y"].shape)
print("pred:", bundle_selected["pred"].shape)
print("t_ref:", bundle_selected["meta"]["t_ref"][0])

# 이후부터는 기존 figure 코드에서
# sample_idx=0, bundle=bundle_selected 로 그대로 사용하면 됨.
bundle = bundle_selected
sample_idx = 0
interpret = bundle["interpret"]
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_dataset_scale_contribution_by_lead(interpret):
    scales = sorted(interpret["scale_outputs"].keys())
    arr0 = interpret["scale_outputs"][scales[0]]
    n_samples, lead_count = arr0.shape[0], arr0.shape[1]
    leads = np.arange(1, lead_count + 1)

    mean_abs_dict = {s: [] for s in scales}
    std_dict = {s: [] for s in scales}
    ratio_dict = {s: [] for s in scales}

    for lead_idx in range(lead_count):
        scale_mean_abs_this_lead = []

        # 먼저 각 scale의 dataset 평균 mean(abs(y_s)) 계산
        for s in scales:
            y = interpret["scale_outputs"][s][:, lead_idx, 0]   # (N, H, W)
            if hasattr(y, "detach"):
                y = y.detach().cpu().numpy()

            # sample별 mean(abs(.))
            sample_mean_abs = np.mean(np.abs(y), axis=(1, 2))   # (N,)
            # sample별 std(.)
            sample_std = np.std(y, axis=(1, 2))                 # (N,)

            mean_abs_avg = sample_mean_abs.mean()
            std_avg = sample_std.mean()

            mean_abs_dict[s].append(mean_abs_avg)
            std_dict[s].append(std_avg)
            scale_mean_abs_this_lead.append(mean_abs_avg)

        # ratio 계산
        scale_mean_abs_this_lead = np.array(scale_mean_abs_this_lead, dtype=np.float64)
        ratios = scale_mean_abs_this_lead / (scale_mean_abs_this_lead.sum() + 1e-12)

        for i, s in enumerate(scales):
            ratio_dict[s].append(ratios[i])

    # ---------------------------
    # 1) mean(abs(y_s))
    # ---------------------------
    plt.figure(figsize=(8, 5))
    for s in scales:
        plt.plot(leads, mean_abs_dict[s], marker='o', label=f'scale {s}')
    plt.xlabel("Lead time")
    plt.ylabel("Dataset-avg Mean(|y_s|)")
    plt.title("Scale contribution magnitude by lead time (dataset average)")
    plt.xticks(leads)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.ylim(0, 1.0)
    plt.show()

    # ---------------------------
    # 3) std(y_s)
    # ---------------------------
    plt.figure(figsize=(8, 5))
    for s in scales:
        plt.plot(leads, std_dict[s], marker='o', label=f'scale {s}')
    plt.xlabel("Lead time")
    plt.ylabel("Dataset-avg Std(y_s)")
    plt.title("Spatial variation by lead time (dataset average)")
    plt.xticks(leads)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.ylim(0, 0.5)
    plt.show()

    return mean_abs_dict, ratio_dict, std_dict
```

```python
import numpy as np
import matplotlib.pyplot as plt
import torch

# -------------------------------------------------
# 1) 특정 query token의 "전체 key token" attention을 heatmap으로 보기
#    -> shape: (1, N_total)
# -------------------------------------------------
def plot_spatial_attention_full_token_vector(
    interpret,
    sample_idx=0,
    time_idx=None,
    query_token_idx=330,
    cmap="viridis",
):
    block_idx = max(interpret["spatial_attn"].keys())
    attn_s = interpret["spatial_attn"][block_idx]

    if torch.is_tensor(attn_s):
        attn_s = attn_s.cpu().numpy()

    if time_idx is None:
        time_idx = cfg["input_len"] - 1

    # average_attn_weights=True  -> (B, T, N_total, N_total)
    # average_attn_weights=False -> (B, T, Hh, N_total, N_total)
    if attn_s.ndim == 4:
        vec = attn_s[sample_idx, time_idx, query_token_idx]  # (N_total,)
    elif attn_s.ndim == 5:
        vec = attn_s[sample_idx, time_idx, :, query_token_idx].mean(axis=0)  # (N_total,)
    else:
        raise ValueError(f"Unexpected attn_s ndim: {attn_s.ndim}")

    plt.figure(figsize=(12, 2.2))
    im = plt.imshow(vec[None, :], aspect="auto", cmap=cmap, origin="upper")
    plt.title(
        f"Spatial attention over full key-token axis\n"
        f"block={block_idx}, sample={sample_idx}, time={time_idx}, query={query_token_idx}"
    )
    plt.xlabel("Key token index (global)")
    plt.yticks([])
    plt.colorbar(im, fraction=0.025, pad=0.02)
    plt.tight_layout()
    plt.show()


# -------------------------------------------------
# 2) 특정 time에서 전체 spatial attention matrix 보기
#    -> shape: (N_total, N_total)
# -------------------------------------------------
def plot_spatial_attention_full_matrix(
    interpret,
    sample_idx=0,
    time_idx=None,
    cmap="viridis",
    vmax_percentile=None,
):
    block_idx = max(interpret["spatial_attn"].keys())
    attn_s = interpret["spatial_attn"][block_idx]

    if torch.is_tensor(attn_s):
        attn_s = attn_s.cpu().numpy()

    if time_idx is None:
        time_idx = cfg["input_len"] - 1

    if attn_s.ndim == 4:
        mat = attn_s[sample_idx, time_idx]   # (N_total, N_total)
    elif attn_s.ndim == 5:
        mat = attn_s[sample_idx, time_idx].mean(axis=0)  # head 평균 -> (N_total, N_total)
    else:
        raise ValueError(f"Unexpected attn_s ndim: {attn_s.ndim}")

    vmin = 0.0
    if vmax_percentile is None:
        vmax = mat.max()
    else:
        vmax = np.percentile(mat, vmax_percentile)

    plt.figure(figsize=(7, 6))
    im = plt.imshow(mat, aspect="auto", cmap=cmap, origin="upper", vmin=vmin, vmax=vmax)
    plt.title(
        f"Full spatial attention matrix\n"
        f"block={block_idx}, sample={sample_idx}, time={time_idx}"
    )
    plt.xlabel("Key token index (global)")
    plt.ylabel("Query token index (global)")
    plt.colorbar(im, fraction=0.046, pad=0.04)
    plt.tight_layout()
    plt.show()


# -------------------------------------------------
# 3) scale 경계선까지 같이 넣어서 전체 토큰축에서 보기
# -------------------------------------------------
def plot_spatial_attention_full_matrix_with_scale_boundaries(
    interpret,
    sample_idx=0,
    time_idx=None,
    cmap="viridis",
    vmax_percentile=99,
):
    block_idx = max(interpret["spatial_attn"].keys())
    attn_s = interpret["spatial_attn"][block_idx]

    if torch.is_tensor(attn_s):
        attn_s = attn_s.cpu().numpy()

    if time_idx is None:
        time_idx = cfg["input_len"] - 1

    if attn_s.ndim == 4:
        mat = attn_s[sample_idx, time_idx]
    elif attn_s.ndim == 5:
        mat = attn_s[sample_idx, time_idx].mean(axis=0)
    else:
        raise ValueError(f"Unexpected attn_s ndim: {attn_s.ndim}")

    bounds = scale_boundaries_from_shapes(interpret["scale_token_shapes"])
    vmax = np.percentile(mat, vmax_percentile) if vmax_percentile is not None else mat.max()

    plt.figure(figsize=(7.5, 6.5))
    im = plt.imshow(mat, aspect="auto", cmap=cmap, origin="upper", vmin=0.0, vmax=vmax)

    # scale boundary 선
    for s, (st, ed) in bounds.items():
        if st > 0:
            plt.axvline(st - 0.5, color="white", linestyle="--", linewidth=0.8, alpha=0.8)
            plt.axhline(st - 0.5, color="white", linestyle="--", linewidth=0.8, alpha=0.8)

        center = (st + ed - 1) / 2.0
        plt.text(center, -8, f"s={s}", ha="center", va="bottom", fontsize=9, color="black")
        plt.text(-8, center, f"s={s}", ha="right", va="center", fontsize=9, color="black", rotation=90)

    plt.title(
        f"Full spatial attention matrix with scale boundaries\n"
        f"block={block_idx}, sample={sample_idx}, time={time_idx}"
    )
    plt.xlabel("Key token index (global)")
    plt.ylabel("Query token index (global)")
    plt.colorbar(im, fraction=0.046, pad=0.04)
    plt.tight_layout()
    plt.show()


# -------------------------------------------------
# 사용 예시
# -------------------------------------------------

# A. 지금 네가 보던 query token 하나를 전체 token 축 기준 heatmap으로 보기
plot_spatial_attention_full_token_vector(
    interpret=interpret,
    sample_idx=0,
    time_idx=cfg["input_len"] - 1,
    query_token_idx=3,
    cmap="viridis",
)

# B. 특정 시점의 전체 spatial attention matrix 보기
plot_spatial_attention_full_matrix(
    interpret=interpret,
    sample_idx=0,
    time_idx=cfg["input_len"] - 1,
    cmap="viridis",
    vmax_percentile=99,   # 너무 밝은 값 몇 개 때문에 전체가 죽는 거 방지
)

# C. scale 경계선까지 같이 보기
plot_spatial_attention_full_matrix_with_scale_boundaries(
    interpret=interpret,
    sample_idx=0,
    time_idx=cfg["input_len"] - 1,
    cmap="viridis",
    vmax_percentile=99,
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import torch


def get_scale_token_ranges_from_interpret(interpret):
    bounds = {}
    cursor = 0
    for s in sorted(interpret["scale_token_shapes"].keys()):
        _, _, Ns, _ = interpret["scale_token_shapes"][s]
        bounds[int(s)] = (cursor, cursor + Ns)
        cursor += Ns
    return bounds


def pixel_to_local_query_token(h, w, scale, H=32, W=32):
    gh = H // scale
    gw = W // scale
    ph = min(int(h * gh / H), gh - 1)
    pw = min(int(w * gw / W), gw - 1)
    q_local = ph * gw + pw
    return q_local, ph, pw, gh, gw


def upsample_token_map_to_image(token_map, scale):
    """
    token_map: (gh, gw)
    return: (H, W) where H=gh*scale, W=gw*scale
    """
    return np.kron(token_map, np.ones((scale, scale), dtype=np.float32))


def normalize_01(x, eps=1e-12):
    x = np.asarray(x, dtype=np.float32)
    mn = np.nanmin(x)
    mx = np.nanmax(x)
    if mx - mn < eps:
        return np.zeros_like(x)
    return (x - mn) / (mx - mn)


def plot_query_pixel_spatial_attention_overlay(
    bundle,
    sample_idx=0,
    time_idx=None,
    query_h=0,
    query_w=0,
    mean=None,
    std=None,
    bg_cmap="coolwarm",
    attn_cmap="magma",
    attn_alpha=0.65,
    mask=None,
    denorm_background=True,
    use_full_token_axis=False,
):
    """
    bundle: {
        "x": ...,
        "interpret": ...,
        ...
    }

    sample_idx: 배치 내부 샘플 인덱스
    time_idx: 입력 시점 인덱스. None이면 마지막 입력 시점 사용
    query_h, query_w: 원본 32x32 격자에서 query pixel 위치
    mean, std: 배경 SST 역정규화용
    use_full_token_axis=False:
        - 각 scale 내부 key token만 사용하여 overlay
        - 해석상 가장 자연스러움
    use_full_token_axis=True:
        - 전체 token axis에서 해당 scale 구간만 잘라 overlay
        - 현재 False 권장
    """
    interpret = bundle["interpret"]
    x = bundle["x"]

    if time_idx is None:
        time_idx = x.shape[1] - 1

    if torch.is_tensor(x):
        bg = x[sample_idx, time_idx, 0].detach().cpu().numpy()
    else:
        bg = np.asarray(x[sample_idx, time_idx, 0])

    if denorm_background:
        if mean is None or std is None:
            raise ValueError("denorm_background=True 이면 mean, std를 함께 넣어주세요.")
        bg = bg * std + mean

    H, W = bg.shape
    block_idx = max(interpret["spatial_attn"].keys())
    attn_s = interpret["spatial_attn"][block_idx]
    if torch.is_tensor(attn_s):
        attn_s = attn_s.detach().cpu().numpy()

    ranges = get_scale_token_ranges_from_interpret(interpret)
    scales = sorted(ranges.keys())

    fig, axes = plt.subplots(1, len(scales) + 1, figsize=(4.0 * (len(scales) + 1), 4.2))
    if len(scales) == 0:
        axes = [axes]

    # -----------------------------
    # 첫 패널: 원본 SST + query pixel
    # -----------------------------
    bg_show = np.flipud(bg)
    if mask is not None:
        m = np.asarray(mask).astype(bool)
        m = np.flipud(m)
        bg_show = np.ma.masked_where(~m, bg_show)

    im_bg = axes[0].imshow(bg_show, cmap=bg_cmap, origin="upper")
    axes[0].scatter(query_w, H - 1 - query_h, c="lime", s=55, marker="x", linewidths=2)
    axes[0].set_title(f"Query pixel\n(h={query_h}, w={query_w})", fontsize=12)
    axes[0].set_xticks([])
    axes[0].set_yticks([])

    # -----------------------------
    # scale별 overlay
    # -----------------------------
    overlay_last = None

    for ax, scale in zip(axes[1:], scales):
        st, ed = ranges[scale]
        q_local, ph, pw, gh, gw = pixel_to_local_query_token(query_h, query_w, scale, H=H, W=W)

        # query token의 global index
        q_global = st + q_local

        # average_attn_weights=True  -> (B, T, N_total, N_total)
        # average_attn_weights=False -> (B, T, Hh, N_total, N_total)
        if attn_s.ndim == 4:
            if use_full_token_axis:
                vec_full = attn_s[sample_idx, time_idx, q_global]   # (N_total,)
                vec = vec_full[st:ed]                                # 해당 scale key만 사용
            else:
                vec = attn_s[sample_idx, time_idx, q_global, st:ed]  # (Ns_scale,)
        elif attn_s.ndim == 5:
            if use_full_token_axis:
                vec_full = attn_s[sample_idx, time_idx, :, q_global].mean(axis=0)
                vec = vec_full[st:ed]
            else:
                vec = attn_s[sample_idx, time_idx, :, q_global, st:ed].mean(axis=0)
        else:
            raise ValueError(f"Unexpected attn_s ndim: {attn_s.ndim}")

        attn_map_token = vec.reshape(gh, gw)
        attn_map_img = upsample_token_map_to_image(attn_map_token, scale)
        attn_map_img = normalize_01(attn_map_img)

        bg_show = np.flipud(bg)
        ov_show = np.flipud(attn_map_img)

        if mask is not None:
            m = np.asarray(mask).astype(bool)
            m2 = np.flipud(m)
            bg_show = np.ma.masked_where(~m2, bg_show)
            ov_show = np.ma.masked_where(~m2, ov_show)

        ax.imshow(bg_show, cmap=bg_cmap, origin="upper")
        overlay_last = ax.imshow(
            ov_show,
            cmap=attn_cmap,
            origin="upper",
            alpha=attn_alpha * ov_show
        )

        # query pixel marker
        ax.scatter(query_w, H - 1 - query_h, c="cyan", s=50, marker="x", linewidths=2)

        # query token 중심 표시
        token_center_y = (ph + 0.5) * scale - 0.5
        token_center_x = (pw + 0.5) * scale - 0.5
        ax.scatter(token_center_x, H - 1 - token_center_y, c="yellow", s=28, marker="o", edgecolors="black")

        ax.set_title(
            f"Scale {scale}\nquery token=({ph},{pw})",
            fontsize=12
        )
        ax.set_xticks([])
        ax.set_yticks([])

    cbar1 = fig.colorbar(im_bg, ax=axes[0], fraction=0.046, pad=0.04)
    cbar1.set_label("SST", fontsize=10)

    if overlay_last is not None:
        cbar2 = fig.colorbar(overlay_last, ax=axes[-1], fraction=0.046, pad=0.04)
        cbar2.set_label("Normalized attention", fontsize=10)

    fig.suptitle(
        f"Spatial attention overlay | block={block_idx} | sample={sample_idx} | time={time_idx}",
        fontsize=14,
        y=1.02
    )

    plt.tight_layout()
    plt.show()


# =========================================================
# 사용 예시
# - query_h, query_w 를 보고 싶은 픽셀 위치로 바꾸면 됨
# - 보통 마지막 입력 프레임(time_idx = cfg['input_len']-1) 기준
# =========================================================

plot_query_pixel_spatial_attention_overlay(
    bundle=bundle,
    sample_idx=23,
    time_idx=cfg["input_len"] - 1,
    query_h=10,
    query_w=10,
    mean=mean,
    std=std,
    bg_cmap="coolwarm",
    attn_cmap="YlGn",
    attn_alpha=0.55,
    mask=mask,
    denorm_background=True,
    use_full_token_axis=False,
)
```
